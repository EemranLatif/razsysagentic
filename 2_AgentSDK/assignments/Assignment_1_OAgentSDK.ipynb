```python
from dotenv import load_dotenv
from agents import Agent, Runner, trace

load_dotenv(override=True)

agent = Agent(
    name="Travel Concierge",
    instructions=(
        "You are a friendly, knowledgeable travel concierge. "
        "Recommend destinations, build simple day-by-day itineraries, and give practical travel tips."
    ),
    model="gpt-4o-mini",
)

with trace("Travel planning"):
    result = await Runner.run(
        agent,
        "Plan a 3-day trip to Lisbon for a couple who love food and architecture"
    )

print(result.final_output)
```
