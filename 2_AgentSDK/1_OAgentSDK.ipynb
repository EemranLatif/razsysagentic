# OpenAI Agent SDK — First Steps

This notebook introduces the three core building blocks of the OpenAI Agents SDK:

| Concept | What it does |
|---|---|
| `Agent` | Defines *who* the AI is — its name, System prompt(instructions), and model |
| `Runner` | Runs the agent against a prompt and returns a result |
| `trace` | Wraps a run so you can inspect it on the OpenAI traces dashboard |

```python
# --- Imports ---
# dotenv   : loads your OPENAI_API_KEY from a local .env file
# Agent    : the AI persona (name + instructions + model)
# Runner   : executes the agent against a prompt
# trace    : wraps a run for observability on platform.openai.com/traces

from dotenv import load_dotenv
from agents import Agent, Runner, trace
```

```python
# --- Load environment variables ---
# Reads .env in the current directory and injects variables into os.environ.
# override=True means values in .env always win over anything already set.

load_dotenv(override=True)
```

```python
# --- Create an Agent ---
# name         : a human-readable label (useful in traces and multi-agent setups)
# instructions : the system prompt — tells the model how to behave
# model        : which OpenAI model to use

agent = Agent(
    name="Financial Analyst",
    instructions="You are a financial analyst. Provide clear, concise, and accurate financial insights.",
    model="gpt-4o-mini",
)
```

```python
# --- Run the Agent ---
# trace()       : groups everything inside it as a single named run in the dashboard
# Runner.run()  : sends the prompt to the agent and waits for a response
# final_output  : the plain-text reply from the model

with trace("Financial analysis"):
    result = await Runner.run(agent, "What are the key metrics to evaluate a company's financial health?")

print(result.final_output)
```

## Check the trace

Open [platform.openai.com/traces](https://platform.openai.com/traces) and you should see a **"Financial analysis"** run.

The trace shows:
- The exact prompt sent to the model
- Token usage and latency
- Every step the agent took (useful once agents start using tools)
