```python
### Without context management (risky)
f = open("data/data.txt", "r")
data = f.read()
# if an error happens here, f.close() never runs — file stays open
print(data)
f.close()
```

```python
### With context management (safe)
# Reading a file
with open("data/data.txt", "r") as f:
    data = f.read()
    print(data)
# file is automatically closed here — even if an error occurred inside
```

```python
with open("data/output.txt", "w") as f:
    f.write("Hello World")
# file is saved and closed automatically
```

```python
# Appending to a file
with open("data/output.txt", "a") as f:
    f.write("\nNew line added")
# file is closed automatically
```
