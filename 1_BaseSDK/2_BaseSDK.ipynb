```python
# And now - the all important import statement
# If you get an import error - head over to troubleshooting in the Setup folder
# Even for other LLM providers like Gemini, you still use this OpenAI import - see Guide 9 for why

from openai import OpenAI
```

```python
# And now we'll create an instance of the OpenAI class
# If you're not sure what it means to create an instance of a class - head over to the guides folder (guide 6)!
# If you get a NameError - head over to the guides folder (guide 6)to learn about NameErrors - always instantly fixable
# If you're not using OpenAI, you just need to slightly modify this - precise instructions are in the AI APIs guide (guide 9)

openai = OpenAI()
```

```python
# And now - let's ask for a question:

question = "pick a business area that might be worth exploring for an Agentic AI opportunity"
messages = [{"role": "user", "content": question}]
```

```python
# ask it - this uses GPT 4.1 mini, still cheap but more powerful than nano

response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)

Agentic_AI_opportunity = response.choices[0].message.content

print(Agentic_AI_opportunity)
```

```python
# form a new messages list
messages = [{"role": "user", "content":  "pain-point in that industry - something challenging that might be ripe for an Agentic solution" + Agentic_AI_opportunity}]
```

```python
# Ask it again

response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)

pain_point = response.choices[0].message.content
print(pain_point)
```

```python
# form a new messages list
messages = [{"role": "user", "content":  "propose the Agentic AI solution for " + pain_point}]
```

```python
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)

answer = response.choices[0].message.content
print(answer)
```
