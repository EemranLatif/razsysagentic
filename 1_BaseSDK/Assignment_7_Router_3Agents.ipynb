In this assignment you will build a **Router Agent** system. A router is the simplest agentic pattern: one model reads the user's input and decides which specialist agent should handle it. The specialist then produces the final answer.

```
User Input
    │
    ▼
┌─────────────┐
│  ROUTER     │  ← decides which agent to use
│  (GPT)      │
└──────┬──────┘
       │
  ┌────┴─────────────┐
  │                  │                  │
  ▼                  ▼                  ▼
Agent A          Agent B           Agent C
Tech Support     Billing           General FAQ
```

````python
# SOLUTION — Task 1
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)
openai = OpenAI()
````

````python
# SOLUTION — Task 2
tech_support_prompt = """
You are an expert technical support engineer. You help users solve software bugs,
error messages, API issues, and technical problems. Be precise and methodical.
If you need more information to diagnose the problem, ask one clarifying question at a time.
Always explain what is likely causing the issue before suggesting a fix.
"""

billing_prompt = """
You are a polite and empathetic billing specialist. You help users with invoices,
payment issues, refunds, subscription charges, and account billing queries.
Always acknowledge the user's concern with empathy before explaining the process.
Be clear about what you can and cannot action directly.
"""

general_faq_prompt = """
You are a friendly and knowledgeable customer assistant. You answer general questions
about the product, pricing plans, free trials, company information, and opening hours.
Keep your answers concise, warm, and helpful. If you don't know the answer, say so
and offer to connect them with the right team.
"""
````

````python
# SOLUTION — Task 3
def router(user_message):
    system = (
        "You are a message classification router for a customer support system. "
        "Read the user's message and classify it into exactly one of these three categories:\n"
        "  tech_support   — software bugs, errors, API issues, technical problems\n"
        "  billing        — invoices, payments, refunds, charges, subscriptions\n"
        "  general_faq    — product questions, pricing, free trials, opening hours, general info\n\n"
        "Respond with ONLY the category label — one word, lowercase, no punctuation, no explanation."
    )
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": system},
            {"role": "user",   "content": user_message}
        ]
    )
    return response.choices[0].message.content.strip().lower()

# Quick test
print(router("My app keeps crashing when I upload a file"))
print(router("I was charged twice on my last invoice"))
print(router("What are your business hours?"))
````

````python
# SOLUTION — Task 4
def tech_support_agent(user_message):
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": tech_support_prompt},
            {"role": "user",   "content": user_message}
        ]
    )
    return response.choices[0].message.content

def billing_agent(user_message):
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": billing_prompt},
            {"role": "user",   "content": user_message}
        ]
    )
    return response.choices[0].message.content

def general_faq_agent(user_message):
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": general_faq_prompt},
            {"role": "user",   "content": user_message}
        ]
    )
    return response.choices[0].message.content
````

````python
# SOLUTION — Task 5
def dispatcher(user_message):
    label = router(user_message)
    print(f"🔀 Router decision: [{label}]")

    if label == "tech_support":
        return tech_support_agent(user_message)
    elif label == "billing":
        return billing_agent(user_message)
    elif label == "general_faq":
        return general_faq_agent(user_message)
    else:
        return "I'm sorry, I wasn't able to identify the right team for your query. Please contact support@example.com."
````

````python
# SOLUTION — Task 6
test_messages = [
    "I keep getting a 403 error when I try to log in via the API",
    "Can I get a refund for last month's subscription?",
    "Do you offer a free trial?",
    "The dashboard is not loading on Chrome but works on Firefox",
    "How do I upgrade my plan?"
]

for msg in test_messages:
    print(f"\n📨 User: {msg}")
    print(dispatcher(msg))
    print("=" * 60)
````

````python
# SOLUTION — Task 7
counts = {"tech_support": 0, "billing": 0, "general_faq": 0}

def dispatcher_with_tracking(user_message):
    label = router(user_message)
    print(f"🔀 Router decision: [{label}]")

    if label == "tech_support":
        counts["tech_support"] += 1
        return tech_support_agent(user_message)
    elif label == "billing":
        counts["billing"] += 1
        return billing_agent(user_message)
    elif label == "general_faq":
        counts["general_faq"] += 1
        return general_faq_agent(user_message)
    else:
        return "I'm sorry, I wasn't able to route your request. Please contact support@example.com."

for msg in test_messages:
    print(f"\n📨 User: {msg}")
    print(dispatcher_with_tracking(msg))
    print("=" * 60)

print("\n📊 Agent usage summary:")
for agent, count in counts.items():
    print(f"  {agent}: {count} call(s)")

# Reflection answers:
# 1. Ambiguous messages — the router picks the closest category. You could add a
#    'confidence' check by asking the router to also return a score, and escalate
#    to a human if confidence is low.
# 2. Robustness — add a validation step that checks the label is one of the three
#    expected values before dispatching; retry once if not.
# 3. Real system additions — conversation history, ticket logging, human escalation,
#    authentication, and a CRM integration.
````
