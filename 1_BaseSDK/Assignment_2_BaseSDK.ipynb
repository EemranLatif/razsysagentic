```python
# SOLUTION — Task 1
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)
openai = OpenAI()
```

```python
# SOLUTION — Task 2
from IPython.display import Markdown, display
```

```python
# SOLUTION — Task 3
question = "What is a promising Agentic AI opportunity in the Real Estate industry? Explain the use case, who benefits, and why it needs an AI agent rather than a simple chatbot."
messages = [{"role": "user", "content": question}]
```

```python
# SOLUTION — Task 4
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)
answer = response.choices[0].message.content
print(answer)
```

```python
# SOLUTION — Task 5
display(Markdown(answer))
```

```python
# SOLUTION — Task 6

# Step 1: add the assistant's reply to the conversation history
messages.append({"role": "assistant", "content": answer})

# Step 2: add the follow-up question
messages.append({"role": "user", "content": "What would be the biggest technical challenge in building this?"})

# Step 3: call the API with the full conversation history
response2 = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)
answer2 = response2.choices[0].message.content

# Step 4: render nicely
display(Markdown(answer2))

# Note: without appending the previous answer to messages, the model would
# have no idea what "this" refers to in the follow-up question!
```
