# Assignment: Agentic Pattern 2 — Parallel Agents with Synthesizer

In this assignment you will build a **Parallel Agent** system. Unlike a router (which picks one agent), parallel agents run at the **same time** and each produces their own output. A synthesizer then combines all outputs into one final, unified response.

```
                User Input
                    │
          ┌─────────┴─────────┐
          │                   │
          ▼                   ▼
    ┌───────────┐       ┌───────────┐
    │  Agent A  │       │  Agent B  │   ← run in parallel (threads)
    │  Optimist │       │  Critic   │
    └─────┬─────┘       └─────┬─────┘
          │                   │
          └────────┬──────────┘
                   │
                   ▼
          ┌─────────────────┐
          │   SYNTHESIZER   │   ← combines both outputs
          └────────┬────────┘
                   │
                   ▼
             Final Answer
```

**You will build:**
- An **Optimist Agent** that finds the strengths and opportunities in any business idea
- A **Critic Agent** that identifies the risks and weaknesses
- Both agents run **in parallel using Python threads** so neither waits for the other
- A **Synthesizer** that reads both outputs and produces a balanced final verdict

### Task 2 — Define the two agent system prompts

Create two string variables:

- `optimist_prompt` — an enthusiastic business analyst who identifies strengths, market opportunities, competitive advantages, and reasons why a business idea could succeed. Always finds the positive angle.

- `critic_prompt` — a sharp-minded risk analyst who identifies weaknesses, market threats, execution challenges, and reasons why a business idea could fail. Always finds the risks and blind spots.

Both should ask the model to structure their response with clear headings.

````python
# YOUR CODE HERE
optimist_prompt = """ you are an optimistic analyst.
you do following tasks:
- you identifies strengths and opportunities in a given situation.
- you provides competitive advantages and encouragement.
- you provides reason why business or project will succeed and how to improve it.
- Always find a positive aspect in any situation.

"""

critic_prompt = """ you are a critical analyst.
you do following tasks:
- you identifies weaknesses of market.
- you analyzes market threats and risks.
- you provides competitive advantages and encouragement.
- you provides reason why business or project will fail.
- you identifies blind spots and areas of improvement.

"""
````

### Task 3 — Write the two agent functions

Write two functions — `optimist_agent(idea, results, index)` and `critic_agent(idea, results, index)`.

Each function:
1. Calls `gpt-4o-mini` with the appropriate system prompt and the `idea` as user message
2. Stores the response text at `results[index]`

> 💡 Why `results[index]`? When using threads, functions can't `return` values directly back to the caller. Instead, we pass in a shared list and each thread writes its result into a specific slot.

````python
def optimist_agent(idea, results, index):
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": optimist_prompt},
            {"role": "user",   "content": idea}
        ]
    )
    results[index] = response.choices[0].message.content

def critic_agent(idea, results, index):
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": critic_prompt},
            {"role": "user",   "content": idea}
        ]
    )
    results[index] = response.choices[0].message.content
````

### Task 4 — Write the synthesizer

Write a function called `synthesizer(idea, optimist_output, critic_output)` that:

1. Builds a prompt that gives the model both the optimist and critic outputs
2. Calls `gpt-4o-mini` with a system prompt that instructs it to synthesize both perspectives into a **balanced, actionable final verdict**
3. The final verdict should include: overall assessment, top 3 strengths, top 3 risks, and a go/no-go recommendation
4. Returns the synthesized response text

````python
def synthesizer(idea, optimist_output, critic_output):
    system = """
    You are a senior business strategist. You have received two analyses of a business idea:
    one from an optimist and one from a critic. Your job is to synthesize both perspectives
    into a single, balanced, actionable verdict.
    
    Structure your response with these four sections:
    ## Overall Assessment
    ## Top 3 Strengths
    ## Top 3 Risks
    ## Go / No-Go Recommendation (with reasoning)
    """
    user_content = (
        f"Business idea: {idea}\n\n"
        f"--- OPTIMIST ANALYSIS ---\n{optimist_output}\n\n"
        f"--- CRITIC ANALYSIS ---\n{critic_output}\n\n"
        f"Now synthesize both perspectives into your balanced verdict."
    )
    response = openai.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": system},
            {"role": "user",   "content": user_content}
        ]
    )
    return response.choices[0].message.content
````

### Task 5 — Run the agents in parallel using threads

Write a function called `run_parallel_analysis(idea)` that:

1. Creates a `results` list with 2 slots: `results = [None, None]`
2. Creates two `threading.Thread` objects — one for `optimist_agent`, one for `critic_agent` — passing `results` and the correct index
3. Starts **both** threads
4. Calls `.join()` on both threads to wait for them to finish
5. Prints both outputs (labelled clearly)
6. Calls `synthesizer()` with both results and returns the final verdict

> 💡 Start both threads **before** calling join on either — this is what makes them run in parallel, not one after the other.

````python
# SOLUTION — Task 5
def run_parallel_analysis(idea):
    print(f"\n🔍 Analysing idea: {idea}")
    print("Running Optimist and Critic agents in parallel...\n")

    results = [None, None]

    t1 = threading.Thread(target=optimist_agent, args=(idea, results, 0))
    t2 = threading.Thread(target=critic_agent,   args=(idea, results, 1))

    # Start BOTH before joining either — this is what makes them parallel
    t1.start()
    t2.start()

    t1.join()
    t2.join()

    print("=" * 60)
    print("🟢 OPTIMIST AGENT OUTPUT:")
    print(results[0])
    print("\n" + "=" * 60)
    print("🔴 CRITIC AGENT OUTPUT:")
    print(results[1])
    print("\n" + "=" * 60)
    print("⚖️  SYNTHESIZER OUTPUT:")
    verdict = synthesizer(idea, results[0], results[1])
    print(verdict)
    return verdict
````

### Task 7 — Measure parallel vs sequential time

Import `time`. Run the two agents **sequentially** (one after the other, without threads), then run them **in parallel** (with threads). Time both approaches using `time.time()` and print the elapsed time for each.

> 💡 This demonstrates the real-world benefit of parallel execution — you should see the parallel version is roughly twice as fast.

````python
import time




# Parallel run


# YOUR CODE HERE — print both times and the speedup

# SOLUTION — Task 7


test_idea = "An AI-powered tutoring app for high school students that adapts to each student's learning style."
results_seq = [None, None]
results_par = [None, None]

# YOUR CODE HERE — call both agents one after the other, time it
# Sequential run — one after the other
start = time.time()
optimist_agent(test_idea, results_seq, 0)
critic_agent(test_idea, results_seq, 1)
seq_time = time.time() - start
print(f"Sequential time: {seq_time:.2f}s")

# Parallel run — both at the same time
# YOUR CODE HERE — call both agents in parallel with threads, time it

start = time.time()
t1 = threading.Thread(target=optimist_agent, args=(test_idea, results_par, 0))
t2 = threading.Thread(target=critic_agent,   args=(test_idea, results_par, 1))
t1.start()
t2.start()
t1.join()
t2.join()
par_time = time.time() - start
print(f"Parallel time:   {par_time:.2f}s")

speedup = seq_time / par_time
print(f"\n⚡ Parallel was {speedup:.1f}x faster")
# Expected: parallel is roughly 2x faster because both API calls run at the same time
````
