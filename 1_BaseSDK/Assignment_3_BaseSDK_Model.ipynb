````python
# SOLUTION — Task 1
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

load_dotenv(override=True)

openai_api_key = os.getenv('OPENAI_API_KEY')
google_api_key = os.getenv('GOOGLE_API_KEY')

print(f"OpenAI: {openai_api_key[:8]}" if openai_api_key else "OpenAI key not set")
print(f"Google: {google_api_key[:4]}" if google_api_key else "Google key not set")
````

````python
# SOLUTION — Task 2
openai = OpenAI()

request = "Come up with a single, open-ended, thought-provoking question about the ethics of Agentic AI in the workplace. Answer only with the question — no preamble, no explanation."
messages = [{"role": "user", "content": request}]

response = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=messages
)
question = response.choices[0].message.content
print(question)
````

````python
# SOLUTION — Task 3
competitors = []
answers = []
messages = [{"role": "user", "content": question}]
````

````python
# SOLUTION — Task 4
response = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=messages
)
competitors.append("gpt-4.1-nano")
answers.append(response.choices[0].message.content)
````

````python
# SOLUTION — Task 5
google_client = OpenAI(
    api_key=os.getenv("GOOGLE_API_KEY"),
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

response = google_client.chat.completions.create(
    model="gemini-2.0-flash",
    messages=messages
)
competitors.append("gemini-2.0-flash")
answers.append(response.choices[0].message.content)
````

````python
# SOLUTION — Task 6
for model, answer in zip(competitors, answers):
    display(Markdown(f"### Response from: `{model}`"))
    display(Markdown(answer))
    print("---")
````

````python
# SOLUTION — Task 7
judge = f"Your task is to rank the following AI responses to this question:\n\n"
judge += f"Question: {question}\n\n"

for i, (model, answer) in enumerate(zip(competitors, answers), start=1):
    judge += f"Response {i} (from {model}):\n{answer}\n\n"

judge += (
    "Rank the responses from best to worst based on depth, clarity, and insight. "
    'Respond ONLY with a JSON object in this exact format: {"results": ["1", "2"]} '
    "where the numbers are the response numbers in ranked order. "
    "Do not include markdown formatting or any explanation."
)

print(judge)
````

````python
# SOLUTION — Task 8
judge_messages = [{"role": "user", "content": judge}]

response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=judge_messages
)
results_raw = response.choices[0].message.content

results_dict = json.loads(results_raw)
ranks = results_dict["results"]

print("\n🏆 Final Rankings:")
for index, result in enumerate(ranks):
    competitor = competitors[int(result) - 1]
    print(f"Rank {index + 1}: {competitor}")

# Reflection answer:
# A human reading 2-3 long answers takes time and can be inconsistent.
# An LLM judge can evaluate all answers against the same criteria in seconds,
# at scale — useful when you're running this pipeline across hundreds of queries.
````
