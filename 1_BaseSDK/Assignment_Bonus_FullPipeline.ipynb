````python
# SOLUTION — Task 1
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from pypdf import PdfReader
import gradio as gr
from IPython.display import Markdown, display

load_dotenv(override=True)
openai = OpenAI()

openai_key = os.getenv('OPENAI_API_KEY')
google_key = os.getenv('GOOGLE_API_KEY')
print(f"OpenAI: {openai_key[:8]}" if openai_key else "OpenAI key not set")
print(f"Google: {google_key[:4]}" if google_key else "Google key not set")
````

````python
# SOLUTION — Task 2
reader = PdfReader("raz/ajaz.pdf")
linkedin = ""
for page in reader.pages:
    text = page.extract_text()
    if text:
        linkedin += text

with open("raz/razsystems.txt", "r", encoding="utf-8") as f:
    summary = f.read()

print("--- LinkedIn (first 300 chars) ---")
print(linkedin[:300])
print("\n--- Summary (first 300 chars) ---")
print(summary[:300])
````

````python
# SOLUTION — Task 3
q_prompt = (
    f"Based on this professional LinkedIn profile, write ONE open-ended question "
    f"a recruiter or potential client might ask this person. "
    f"Return ONLY the question — no preamble, no explanation.\n\n{linkedin}"
)

response = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=[{"role": "user", "content": q_prompt}]
)
recruiter_question = response.choices[0].message.content
print(recruiter_question)
````

````python
# SOLUTION — Task 4
competitors = []
answers = []
messages = [{"role": "user", "content": recruiter_question}]

# Model 1: gpt-4.1-nano
r = openai.chat.completions.create(model="gpt-4.1-nano", messages=messages)
competitors.append("gpt-4.1-nano")
answers.append(r.choices[0].message.content)

# Model 2: gpt-4.1-mini
r = openai.chat.completions.create(model="gpt-4.1-mini", messages=messages)
competitors.append("gpt-4.1-mini")
answers.append(r.choices[0].message.content)

# Model 3: gemini-2.0-flash via Google's OpenAI-compatible endpoint
google_client = OpenAI(
    api_key=os.getenv("GOOGLE_API_KEY"),
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)
r = google_client.chat.completions.create(model="gemini-2.0-flash", messages=messages)
competitors.append("gemini-2.0-flash")
answers.append(r.choices[0].message.content)
````

````python
# SOLUTION — Task 5
for model, answer in zip(competitors, answers):
    display(Markdown(f"## Response from: `{model}`"))
    display(Markdown(answer))
    display(Markdown("---"))
````

````python
# SOLUTION — Task 6
judge = f"Your task is to rank the following AI responses to this question:\n\n"
judge += f"Question: {recruiter_question}\n\n"

for i, (model, answer) in enumerate(zip(competitors, answers), start=1):
    judge += f"Response {i} (from {model}):\n{answer}\n\n"

judge += (
    "Rank the responses from best to worst based on depth, accuracy, and relevance. "
    'Respond ONLY with valid JSON in this exact format: {"results": ["1", "2", "3"]} '
    "No markdown formatting, no explanation."
)

j_response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": judge}]
)
results_raw = j_response.choices[0].message.content
results_dict = json.loads(results_raw)
ranks = results_dict["results"]

ranked_models = [competitors[int(r) - 1] for r in ranks]

print("\n🏆 Final Rankings:")
for i, m in enumerate(ranked_models, 1):
    print(f"  Rank {i}: {m}")
````

````python
# SOLUTION — Task 7
name = "Aiza"   # replace with your own first name

# Find the winning answer by looking up the winner's index in competitors
winning_index = competitors.index(ranked_models[0])
winning_answer = answers[winning_index]

system_prompt = (
    f"You are acting as {name}. You are answering questions on {name}'s personal website, "
    f"particularly questions related to {name}'s career, background, skills and experience. "
    f"Your responsibility is to represent {name} professionally to potential clients or employers. "
    f"If you don't know the answer, say so."
)
system_prompt += f"\n\n## Personal Summary:\n{summary}"
system_prompt += f"\n\n## AI-Generated Career Highlight (by {ranked_models[0]}):\n{winning_answer}"
system_prompt += f"\n\n## LinkedIn Profile:\n{linkedin}"
system_prompt += f"\n\nAlways stay in character as {name}."

print(system_prompt[:500])
````

````python
# SOLUTION — Task 8
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}]
    messages += history
    messages += [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
    return response.choices[0].message.content

gr.ChatInterface(
    fn=chat,
    title=f"{name} — AI Career Assistant (Powered by {ranked_models[0]})"
).launch()
````

````python
# SOLUTION — Task 9
answers_text = "\n\n".join(
    f"{competitors[i]}:\n{answers[i]}" for i in range(len(competitors))
)

report_prompt = f"""Write a short 3-section Markdown report analysing the following AI model competition.

Question asked: {recruiter_question}

Rankings (best to worst): {ranked_models}

All answers:
{answers_text}

Include exactly these three sections:
## 1. Why {ranked_models[0]} Won
## 2. What the Question Reveals About This Professional
## 3. One Recommendation to Improve the LinkedIn PDF
"""

r = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{"role": "user", "content": report_prompt}]
)
report = r.choices[0].message.content
display(Markdown(report))
````
